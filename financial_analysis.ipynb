## Helper Function: CAGR CalculationWe define a reusable function to calculate the Compound Annual Growth Rate (CAGR), which normalizes growth across different time periods.

```python
def calculate_cagr(start, end, years):    """    Calculate Compound Annual Growth Rate.    Parameters:        start (float): Starting value        end (float): Ending value        years (int): Number of years    Returns:        float: CAGR as a percentage    """    if start == 0:        return None    return ((end / start) ** (1 / years) - 1) * 100# Test the functionprint("CAGR Test: $100 → $144.72 over 2 years =", calculate_cagr(100, 144.72, 2), "%")
```

## Executive SummaryThis analysis evaluates the annual financial performance of **Apple, Microsoft, and Tesla** using standardized data extracted from SEC 10-K filings for fiscal years 2023 through 2025. The dataset encompasses revenue, net income, total assets, total liabilities, operating cash flow, growth rates, and debt ratios. The findings reveal three distinctly different strategic trajectories across the peer group.**Microsoft** emerges as the unequivocal growth leader. Over the two-year period, Microsoft achieved a revenue CAGR of **15.30%** and a net income CAGR of **18.63%**, the highest among all three companies. Its asset base expanded by **50.25%** to $619 billion, reflecting aggressive capital deployment in cloud infrastructure, artificial intelligence, and strategic acquisitions. Importantly, this growth was accompanied by balance sheet strengthening: the debt ratio declined from 49.94% to **44.51%**, demonstrating that Microsoft is scaling efficiently without overleveraging. Net margin improved steadily from 34.15% to **36.15%**, establishing Microsoft as the most profitable operator in the cohort on a margin basis.**Apple** presents a contrasting profile of maturity and stability. Revenue grew at a modest **4.20% CAGR**, consistent with a business operating at global scale with limited new market penetration. Net income rebounded strongly in 2025 (+19.53% YoY) after a slight contraction in 2024, lifting the net margin to **26.91%**. Apple's asset base grew only 1.87% over two years, indicating disciplined capital allocation and a focus on shareholder returns rather than expansion. However, Apple's debt ratio of **79.48%** remains the highest in the group, though it improved from 84.38% in 2024. This elevated leverage is a known structural feature of Apple's capital strategy, which relies heavily on debt to fund share buybacks and dividends.**Tesla's** financial trajectory is the most concerning. Revenue contracted by **2.01%** over two years, with a negative CAGR of -1.01%. More alarming is the collapse in profitability: net income fell **74.26%** from $14.97 billion to $3.86 billion, and the net margin eroded from 15.47% to just **4.07%**. This sustained deterioration — spanning both 2024 and 2025 — suggests systemic margin compression rather than a one-time event. Tesla's asset base grew 29.25% during this period, but this capital expansion has not translated into revenue or profit growth, signaling declining capital efficiency. Notably, Tesla is the only company in the dataset with operating cash flow data, which remained stable at approximately $14.7 billion, suggesting non-cash charges may be contributing to the earnings decline.From a comparative standpoint, Microsoft holds the strongest balance sheet (lowest debt ratio, highest equity base of $343.5 billion), the fastest growth, and the highest profitability. Apple offers stability and elite margins but limited growth momentum. Tesla faces a profitability crisis that demands immediate strategic attention. **Missing operating cash flow data for Apple and Microsoft is a material limitation** that should be addressed in future data collection cycles to enable a complete cash flow analysis.
